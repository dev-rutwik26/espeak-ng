```python
content = []
remove_words = []
# Step 1: Load words to remove
with open("common_word_cleaning.txt") as rm:
    for word in rm:
        words = word.split()
        if words:
            content.append(words[0])
with open("remove_names.txt") as rm:
    for word in rm:
        words = word.split()
        if words:
            remove_words.append(words[0])
with open("rutwik_extra", "r") as f, open("rutwik_extra_cleaned.txt", "w") as out:
    for line in f:
        parts = line.split()

        if not parts:
            continue  # skip empty lines

        # Case 1: short words
        if len(parts[0]) < 4:
            cleaned_line = ""   # remove entire line

        # Case 2: word exists in remove list
        elif parts[0] in content:
            cleaned_line = ""   # remove entire line
        elif parts[0] in remove_words:
            cleaned_line = ""
        else:
            cleaned_line = line  # keep original line

        if cleaned_line:
            out.write(cleaned_line)
```
